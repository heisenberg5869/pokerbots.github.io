# 06 ? Abstraction and endgame solving

This notebook constructs histogram distances and action translations. Its examples are small enough to inspect by hand; they do not claim full-game strategy quality.

```python
from pathlib import Path
import sys
for parent in (Path.cwd(), *Path.cwd().parents):
    if (parent / 'pokerbot').is_dir():
        sys.path.insert(0, str(parent))
        break
else:
    raise RuntimeError('Run this notebook inside the pokerbots repository')
```

```python
from pokerbot.abstraction import earth_movers_distance, k_medoids, translation_probabilities, translate_bet

made = [0.10, 0.25, 0.35, 0.20, 0.10]
draw = [0.30, 0.05, 0.10, 0.15, 0.35]
print("mean indices", sum(i*x for i,x in enumerate(made)), sum(i*x for i,x in enumerate(draw)))
print("EMD", earth_movers_distance(made, draw))
```

The two distributions share a mean but differ in shape. EMD accumulates how much probability mass must cross each boundary between bins.

```python
hands = [
    [1, 0, 0, 0], [0.8, 0.2, 0, 0],
    [0, 0, 0.2, 0.8], [0, 0, 0, 1],
]
labels, medoids = k_medoids(hands, 2)
print("cluster labels:", labels)
print("representative hand indices:", medoids)
for i, h in enumerate(hands):
    print(i, "->", labels[i], "distance", earth_movers_distance(h, hands[medoids[labels[i]]]))
```

```python
menu = [0.5, 1.0]
for bet in [0.5, 0.6, 0.72, 0.9, 1.0]:
    print(bet, translation_probabilities(bet, menu), "linear expected:", sum(s*p for s,p in translation_probabilities(bet, menu).items()))
```

Nearest-size translation is deterministic and creates a threshold. Linear randomized translation preserves the off-tree bet size in expectation. This is a translation property, not a safe-solving guarantee. The module lesson discusses the missing opponent opt-out constraint in safe subgame solving.
